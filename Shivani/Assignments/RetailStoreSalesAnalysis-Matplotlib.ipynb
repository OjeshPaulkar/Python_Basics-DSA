```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('retail_store_data.csv')
df.head(5)
```

```python
#### Productwise Sales - To Observe Revenue over different Products.(Line Chart Presentation)


daily_sales = df.groupby('Product')['Sales'].sum()
print(daily_sales)

plt.figure(figsize=(8,4))
plt.plot(daily_sales.index, daily_sales.values, color='black', marker='o')
plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.title('Productwise Sales')
plt.show()



```

```python
#### Productwise Sales - To Observe Revenue over different Products.(PIE presentation)

colors=('magenta','yellowgreen','pink','blue','lightskyblue','orange')
explode=(0,0,0.1,0,0,0)
plt.pie(daily_sales,explode=explode,labels=daily_sales.index,autopct='%1.1f%%',colors=colors)
```

```python
#### Scatter plot between Quantity and Sales to observe how order size affects revenue.

plt.scatter(df['Sales'],df['Quantity'],color='purple',marker='o')
plt.xlabel('Sales')
plt.ylabel('Quantity')
plt.title('Sales Quantity Analysis')
plt.show()
```

```python
#### Visualization of the distribution of sales values.

plt.hist(df['Sales'],bins=10,color='yellowgreen',edgecolor='green')
```

```python
#### Daywise Sales Analysis.

prod_sale=df.groupby('DayOfWeek')['Sales'].sum()
print(prod_sale)

plt.figure(figsize=(8,4))
plt.bar(prod_sale.index,prod_sale.values,color='silver',edgecolor='purple')
```
