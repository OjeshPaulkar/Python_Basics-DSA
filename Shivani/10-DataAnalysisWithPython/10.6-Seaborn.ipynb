##### What is Seaborn?
1. Seaborn is a Python data visualization library built on top of Matplotlib.
2. It provides high-level functions to create attractive, informative, and complex plots with just a few lines of code.
3. It works seamlessly with pandas DataFrames.


### Why use Seaborn over Matplotlib?
✅ Pre-defined beautiful styles and color palettes
✅ Less code for complex plots (heatmaps, pair plots, categorical plots, etc.)
✅ Better integration with DataFrames
✅ Handles statistical visualization easily

```python
import seaborn as sns
```

## sns.load_dataset() : Loads a sample dataset called “tips” that comes built-in with Seaborn.

## Plotting with Seaborn: 
A. Distribution Plots:
1. sns.histplot() → Histogram
2. sns.kdeplot() → Kernel density (smooth curve)
3. sns.distplot() (older, now replaced by histplot + kdeplot)

B. Categorical Plots:
1. sns.barplot() → Average values of categories
2. sns.countplot() → Count of each category
3. sns.boxplot() → Shows distribution with outliers
4. sns.violinplot() → Combines boxplot + KDE

C. Relational Plots:
1. sns.scatterplot() → Scatter plot
2. sns.lineplot() → Line chart

D. Matrix Plots:
1. sns.heatmap() → Correlation matrix, heatmaps

E. Multi-variable Analysis:
1. sns.pairplot() → Pairwise relationships between columns
2. sns.jointplot() → Scatter + distribution

```python
tips=sns.load_dataset('tips')
tips
```

```python
## Relational Plots:
## 1. Scatter Plot

import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill', y='tip', data=tips, color='red')
plt.title('Scattered Plot of Total Bill vs Tips')
plt.show()
```

```python
## 2. Line Plot
sns.lineplot(x='total_bill', y='tip', data=tips,color='lightskyblue')
plt.title('Line Plot of Total Bill and Tip')
plt.show()
```

```python
## Categorical Plots:
## 1. Bar Plots:Total bills per day

sns.barplot(x='day',y='total_bill',data=tips,color='greenyellow',estimator=sum,width=0.6)
plt.title('Bar Chart of Total Bill per Day')
plt.show()
```

```python
## 2. Box Plot

sns.boxplot(x='day',y='total_bill',data=tips,color='pink')    ### The dots above are nothing but outliers.
```

```python
## 3. Voilin Plot:

sns.violinplot(x='day',y='total_bill',data=tips,color='pink',linecolor='purple')
```

```python
## Count Plot: No. of males and females arrived on each day

sns.countplot(x='day', hue='sex', data=tips, palette=['lightgreen', 'green'])
```

```python
## Distribution Plots:
## 1. Histograms

sns.histplot(x='total_bill',bins=10,data=tips,color='blue',kde=True)
plt.show()
```

```python
## 2. KWE Plot:
sns.kdeplot(x='total_bill',data=tips,color='lightskyblue',fill=True)
```

```python
## Multi variable analysis:
## 1. Pair Plots

sns.pairplot(tips[['total_bill', 'tip', 'size', 'sex']])    ## Comparing 4 Variables at the same time.
```

```python
## Matrix Plots:
## 1. Heat Maps : A heatmap is a graphical representation of data where values are represented by colors.
#                 Darker or brighter colors indicate higher or lower values, depending on the color palette.  
## Syntax : sns.heatmap(data, 
#            annot=True,      # show numbers in cells
#            fmt=".2f",       # number format
#            cmap="coolwarm", # color palette
#            linewidths=0.5)  # lines between cells


corr=tips[['total_bill','tip','size']].corr()
corr
sns.heatmap(corr,annot=True)
```
